```python
import pandas as pd

df = pd.read_csv(
    '/kaggle/input/gun-violence-data/gun-violence-data_01-2013_03-2018.csv'
)

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
import missingno
missingno.matrix(df)
```

```python
df=df.iloc[:,:7]
df.head()
```

```python
df.drop(columns=['incident_id','city_or_county','address'],inplace=True)
```

```python
df.loc[:,["date","state","n_killed","n_injured"]].info()
```

```python
#feature engineering
#feature engineering with time series data
#import library for datetime
from datetime import datetime as dt 
df['date']=pd.to_datetime(df['date']) #convert date to datetime format 
print(df['date'].dtype) 
df['year']=df['date'].dt.year
df['month']=df['date'].dt.month
df['day']=df['date'].dt.day
df['day_of_week']=df['date'].dt.dayofweek
#feature engineering
df['total_victim']=df['n_killed'] + df['n_injured'] # sum up the number of deaths and injuries
df.head()
```

```python

#replace numeric day of week values with text 
df['day_of_week']=df['day_of_week'].map({0:'Monday', 1:'Tuesday', 2:'Wednesday',3:'Thursday',4:'Friday',5:'Saturday',6:'Sunday'})
df.head()
```

```python
df['day_of_week'].value_counts()
```

```python
df.groupby('day_of_week')['total_victim'].count().sort_values(ascending=False)
```

```python
df.groupby("day_of_week")["total_victim"].sum().sort_values(ascending=False)
```

```python
df['state'].value_counts()
```

```python
df.groupby('state').size()
```

```python
#the total number of victims of violence in each US state
#first we group the number of victims by state then sum them all up

df.groupby('state')['total_victim'].sum().sort_values(ascending=False)
```

```python
df.groupby('month')['total_victim'].max().sort_values(ascending=False)
```

```python
import numpy as np
df.groupby('month')['total_victim'].agg([np.mean,min,max]).sort_values(by="max",ascending=False)
```

```python
import matplotlib.pyplot as plt
yearly = df.groupby('year')['total_victim'].sum()
yearly.plot(kind='line', title='Total Victims Over Years')
plt.show()
```

```python
#number of gun violence cases in each month from lowest to highest 
df=df.groupby(['state','month','day_of_week'])['total_victim'].sum().sort_values(ascending=False)
df.head(20)
```
